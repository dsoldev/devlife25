## Função find

```python
doce = "pamonhas pamonhas pamonhas pamonhas pamonhas de Piracicaba"
```

```python
doce.find("pamonhas", len(doce) - 1)
```

```python
pamonhas = []

start = 0
palavra = "pamonhas"

not_encontrou = False

while not_encontrou == False:
    pos = doce.find(palavra, start)
    if pos == -1:
        not_encontrou = True
    else:
       start = pos + 1
       pamonhas.append(pos)

print(pamonhas)


```
